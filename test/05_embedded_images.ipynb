# G1 RSS output — embedded images

Demonstrates embedding generated charts and article images into the notebook (outputs are kept).

```python

from pathlib import Path
import json
def find_data_file():
    for c in [Path("g1_articles.jsonl"), Path("../g1_articles.jsonl"), Path("../../g1_articles.jsonl")]:
        if c.exists(): return c
    for p in Path.cwd().rglob("g1_articles.jsonl"): return p
    raise FileNotFoundError("g1_articles.jsonl not found")
DATA_FILE = find_data_file()
print("Using data file:", DATA_FILE.resolve())

import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

rows = [json.loads(line) for line in DATA_FILE.read_text(encoding="utf-8").splitlines() if line.strip()]
df = pd.DataFrame(rows)

fig, ax = plt.subplots(figsize=(8, 4))
df["source_feeds"].apply(len).value_counts().sort_index().plot(kind="bar", ax=ax, color="#9467bd")
ax.set_title("Source feeds per article (embedded output)")
ax.set_xlabel("Number of feeds")
ax.set_ylabel("Articles")
plt.tight_layout()
print("Figure created with", len(fig.axes), "axes")
```
