# G1 RSS output — dataset overview

Interactive exploration of `g1_articles.jsonl` produced by `scrape_g1_rss.py`.

```python

from pathlib import Path
import json

def find_data_file():
    candidates = [Path("g1_articles.jsonl"), Path("../g1_articles.jsonl"), Path("../../g1_articles.jsonl")]
    for c in candidates:
        if c.exists():
            return c
    for p in Path.cwd().rglob("g1_articles.jsonl"):
        return p
    raise FileNotFoundError("g1_articles.jsonl not found")

DATA_FILE = find_data_file()
print("Using data file:", DATA_FILE.resolve())

import pandas as pd
rows = [json.loads(line) for line in DATA_FILE.read_text(encoding="utf-8").splitlines() if line.strip()]
df = pd.DataFrame(rows)
print("shape:", df.shape)
print("columns:", list(df.columns))
print()
print(df.head(10).to_string(index=False))
```

```python

from pathlib import Path
import json

def find_data_file():
    candidates = [Path("g1_articles.jsonl"), Path("../g1_articles.jsonl"), Path("../../g1_articles.jsonl")]
    for c in candidates:
        if c.exists():
            return c
    for p in Path.cwd().rglob("g1_articles.jsonl"):
        return p
    raise FileNotFoundError("g1_articles.jsonl not found")

DATA_FILE = find_data_file()
print("Using data file:", DATA_FILE.resolve())

import pandas as pd
rows = [json.loads(line) for line in DATA_FILE.read_text(encoding="utf-8").splitlines() if line.strip()]
df = pd.DataFrame(rows)
print("Published range:")
print("  min:", df["published"].min())
print("  max:", df["published"].max())
print("  null published:", df["published"].isna().sum())
print()
print("Unique titles:", df["title"].nunique())
print("Unique links:", df["link"].nunique())
```
